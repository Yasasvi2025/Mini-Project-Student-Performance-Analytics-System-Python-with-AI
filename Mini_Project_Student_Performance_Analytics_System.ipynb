# Student Performance Analytics System

**Python with AI – Mini Project**

**Project Title:** Student Performance Analytics System

**Technologies Used:** Python, NumPy, Pandas

**Concepts Covered:** Variables, Data Types, Conditional Statements, Loops, Functions, NumPy, Pandas, Data Analysis, Data Visualization, and Statistical Analysis.

**Project Description:**

The Student Performance Analytics System is a Python-based mini project designed to analyze student academic performance using marks, attendance, and department information. It calculates total and average marks, assigns grades, identifies pass/fail results, analyzes subject-wise and department-wise performance, and identifies top-performing students. The project uses NumPy and Pandas to process data and generate meaningful insights from the student dataset.

## **Code 1: Import Libraries and Load Dataset**

```python
# ================================================================
# CODE 1: IMPORT LIBRARIES AND LOAD DATASET
# ================================================================

import os
import glob
import numpy as np
import pandas as pd
from IPython.display import display

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)

# Search common dataset names in the current folder and Google Colab folder.
possible_names = [
    "student.csv",
    "students.csv",
    "student(1).csv",
    "student_data.csv",
    "students_data.csv",
]
search_dirs = [os.getcwd(), "/content"]
possible_files = [
    os.path.join(folder, name)
    for folder in search_dirs
    for name in possible_names
]

CSV_FILE = next(
    (path for path in possible_files if os.path.isfile(path)),
    None
)

# If a common filename was not found, look for another CSV in the folder,
# excluding the report generated by this notebook.
if CSV_FILE is None:
    other_csv_files = sorted(
        path for folder in search_dirs if os.path.isdir(folder)
        for path in glob.glob(os.path.join(folder, "*.csv"))
        if os.path.basename(path).lower() != "student_performance_analysis.csv"
    )
    if other_csv_files:
        CSV_FILE = other_csv_files[0]

# If no CSV is found, ask the user to upload one in Google Colab.
if CSV_FILE is None:
    try:
        from google.colab import files
        print("No student CSV found. Please upload your student dataset:")
        uploaded = files.upload()
        csv_files = [name for name in uploaded if name.lower().endswith(".csv")]
        if not csv_files:
            raise FileNotFoundError("Please upload a CSV file.")
        CSV_FILE = csv_files[0]
    except ImportError as exc:
        raise FileNotFoundError(
            "No CSV dataset found. Place student.csv, students.csv, "
            "student(1).csv, or another student dataset CSV in the notebook folder."
        ) from exc

print("CSV file selected:", os.path.basename(CSV_FILE))
data = pd.read_csv(CSV_FILE)

print("STUDENT PERFORMANCE ANALYTICS SYSTEM")
print("-" * 60)
print("Dataset loaded successfully!")
print("Number of records:", len(data))
print("Number of columns:", len(data.columns))

display(data)
```

## **Code 2: Validate the Dataset**

```python
# ================================================================
# CODE 2: DATASET VALIDATION
# ================================================================

required_columns = [
    "Student_ID", "Name", "Department",
    "Python", "SQL", "HTML", "CSS",
    "JavaScript", "Attendance"
]

missing_columns = [
    col for col in required_columns
    if col not in data.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

if data.empty:
    raise ValueError("The dataset is empty.")

if data["Student_ID"].isna().any():
    raise ValueError("Student_ID contains missing values.")

if data["Student_ID"].duplicated().any():
    raise ValueError("Duplicate Student_ID values found.")

if data["Name"].isna().any() or data["Department"].isna().any():
    raise ValueError("Name or Department contains missing values.")

subjects = [
    "Python", "SQL", "HTML", "CSS", "JavaScript"
]

for column in subjects + ["Attendance"]:
    data[column] = pd.to_numeric(
        data[column], errors="raise"
    )

if data[subjects + ["Attendance"]].isna().any().any():
    raise ValueError(
        "Missing marks or attendance values found."
    )

for column in subjects + ["Attendance"]:
    if not data[column].between(0, 100).all():
        raise ValueError(
            f"{column} must contain values between 0 and 100."
        )

print("Dataset validation completed successfully.")
print("Number of students:", len(data))
print("Number of subjects:", len(subjects))
print("Missing values:", int(data[required_columns].isna().sum().sum()))

print("\nDataset information:")
data.info()

print("\nStatistical summary:")
display(data[subjects + ["Attendance"]].describe().round(2))
```

## **Code 3: Define Reusable Functions**

```python
# ================================================================
# CODE 3: REUSABLE FUNCTIONS
# ================================================================

PASS_MARK = 40

def calculate_total(row):
    """Calculate total marks across all subjects."""
    return sum(row[subject] for subject in subjects)


def calculate_average(row):
    """Calculate average marks."""
    return calculate_total(row) / len(subjects)


def assign_grade(average):
    """Assign a grade based on average marks."""
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


def check_result(row):
    """Pass if every subject mark is at least 40 and attendance is at least 75%."""
    for subject in subjects:
        if row[subject] < PASS_MARK:
            return "Fail"
    if row["Attendance"] < 75:
        return "Fail"
    return "Pass"


def get_performance_level(average):
    """Classify overall student performance."""
    if average >= 90:
        return "Excellent"
    elif average >= 75:
        return "Very Good"
    elif average >= 60:
        return "Good"
    elif average >= 40:
        return "Average"
    else:
        return "Needs Improvement"


def show_section(title):
    """Display a formatted section heading."""
    print("\n" + "=" * 70)
    print(title)
    print("=" * 70)


print("All reusable functions defined successfully.")
```

## **Code 4: Calculate Student Totals, Averages, Grades and Results**

```python
# ================================================================
# CODE 4: STUDENT PERFORMANCE CALCULATIONS
# ================================================================

analysis = data.copy()

analysis["Total_Marks"] = analysis.apply(
    calculate_total, axis=1
)

analysis["Average_Marks"] = analysis.apply(
    calculate_average, axis=1
)

analysis["Grade"] = analysis["Average_Marks"].apply(
    assign_grade
)

analysis["Result"] = analysis.apply(
    check_result, axis=1
)

analysis["Performance"] = analysis["Average_Marks"].apply(
    get_performance_level
)

analysis["Average_Marks"] = analysis["Average_Marks"].round(2)

show_section("STUDENT PERFORMANCE RESULTS")

result_columns = [
    "Student_ID", "Name", "Department",
    "Total_Marks", "Average_Marks",
    "Grade", "Result", "Attendance", "Performance"
]

display(
    analysis[result_columns]
    .sort_values("Student_ID")
    .reset_index(drop=True)
)
```

## **Code 5: Overall Class Performance**

```python
# ================================================================
# CODE 5: OVERALL CLASS PERFORMANCE
# ================================================================

total_students = len(analysis)

class_average = analysis["Average_Marks"].mean()
highest_average = analysis["Average_Marks"].max()
lowest_average = analysis["Average_Marks"].min()

highest_student = analysis.loc[
    analysis["Average_Marks"].idxmax()
]

lowest_student = analysis.loc[
    analysis["Average_Marks"].idxmin()
]

total_passed = int(
    (analysis["Result"] == "Pass").sum()
)

total_failed = int(
    (analysis["Result"] == "Fail").sum()
)

pass_percentage = total_passed / total_students * 100
fail_percentage = total_failed / total_students * 100

show_section("OVERALL CLASS PERFORMANCE")

print("Total students:", total_students)
print(f"Class average: {class_average:.2f}")
print(f"Highest average: {highest_average:.2f}")
print(f"Lowest average: {lowest_average:.2f}")

print(
    f"Top performer: {highest_student['Name']} "
    f"(ID: {highest_student['Student_ID']})"
)

print(
    f"Lowest performer: {lowest_student['Name']} "
    f"(ID: {lowest_student['Student_ID']})"
)

print("Students passed:", total_passed)
print("Students failed:", total_failed)
print(f"Pass percentage: {pass_percentage:.2f}%")
print(f"Fail percentage: {fail_percentage:.2f}%")
```

## **Code 6: Subject-Wise Performance Analysis**

```python
# ================================================================
# CODE 6: SUBJECT-WISE PERFORMANCE ANALYSIS
# ================================================================

subject_summary = pd.DataFrame({
    "Subject": subjects,
    "Average_Marks": [
        data[subject].mean() for subject in subjects
    ],
    "Highest_Marks": [
        data[subject].max() for subject in subjects
    ],
    "Lowest_Marks": [
        data[subject].min() for subject in subjects
    ],
    "Students_Passed": [
        int((data[subject] >= PASS_MARK).sum())
        for subject in subjects
    ],
    "Students_Failed": [
        int((data[subject] < PASS_MARK).sum())
        for subject in subjects
    ]
})

subject_summary["Average_Marks"] = (
    subject_summary["Average_Marks"].round(2)
)

show_section("SUBJECT-WISE PERFORMANCE")

subject_summary = subject_summary.sort_values(
    "Average_Marks", ascending=False
).reset_index(drop=True)

display(subject_summary)

best_subject = subject_summary.iloc[0]
weakest_subject = subject_summary.iloc[-1]

print(
    f"Best subject: {best_subject['Subject']} "
    f"({best_subject['Average_Marks']:.2f})"
)

print(
    f"Lowest-average subject: {weakest_subject['Subject']} "
    f"({weakest_subject['Average_Marks']:.2f})"
)
```

## **Code 7: Identify Top Five Students**

```python
# ================================================================
# CODE 7: TOP FIVE PERFORMERS
# ================================================================

show_section("TOP FIVE PERFORMING STUDENTS")

top_students = analysis.sort_values(
    by=["Average_Marks", "Total_Marks"],
    ascending=[False, False]
).head(5)

display(
    top_students[
        [
            "Student_ID", "Name", "Department",
            "Total_Marks", "Average_Marks",
            "Grade", "Result"
        ]
    ].reset_index(drop=True)
)
```

## **Code 8: Department-Wise Performance**

```python
# ================================================================
# CODE 8: DEPARTMENT-WISE PERFORMANCE
# ================================================================

department_summary = analysis.groupby("Department").agg(
    Student_Count=("Student_ID", "count"),
    Average_Marks=("Average_Marks", "mean"),
    Average_Attendance=("Attendance", "mean"),
    Students_Passed=(
        "Result", lambda x: (x == "Pass").sum()
    ),
    Students_Failed=(
        "Result", lambda x: (x == "Fail").sum()
    )
).reset_index()

department_summary["Average_Marks"] = (
    department_summary["Average_Marks"].round(2)
)

department_summary["Average_Attendance"] = (
    department_summary["Average_Attendance"].round(2)
)

show_section("DEPARTMENT-WISE PERFORMANCE")

department_summary = department_summary.sort_values(
    "Average_Marks", ascending=False
).reset_index(drop=True)

display(department_summary)

best_department = department_summary.iloc[0]

print(
    f"Highest-average department: "
    f"{best_department['Department']} "
    f"({best_department['Average_Marks']:.2f})"
)
```

## **Code 9: Grade Distribution**

```python
# ================================================================
# CODE 9: GRADE DISTRIBUTION
# ================================================================

grade_order = ["A+", "A", "B", "C", "D", "F"]

grade_counts = (
    analysis["Grade"]
    .value_counts()
    .reindex(grade_order, fill_value=0)
)

grade_summary = pd.DataFrame({
    "Grade": grade_order,
    "Number_of_Students": [
        int(grade_counts[grade]) for grade in grade_order
    ]
})

grade_summary["Percentage"] = (
    grade_summary["Number_of_Students"]
    / total_students * 100
).round(2)

show_section("GRADE DISTRIBUTION")

display(grade_summary)
```

## **Code 10: Attendance Analysis**

```python
# ================================================================
# CODE 10: ATTENDANCE ANALYSIS
# ================================================================

show_section("ATTENDANCE ANALYSIS")

average_attendance = data["Attendance"].mean()

attendance_eligible = int(
    (data["Attendance"] >= 75).sum()
)

attendance_below_75 = int(
    (data["Attendance"] < 75).sum()
)

print(f"Average attendance: {average_attendance:.2f}%")
print(f"Students with attendance >= 75%: {attendance_eligible}")
print(f"Students with attendance < 75%: {attendance_below_75}")

print("\nStudents with attendance below 75%:")

low_attendance = analysis.loc[
    analysis["Attendance"] < 75,
    ["Student_ID", "Name", "Department", "Attendance"]
]

if low_attendance.empty:
    print("No students have attendance below 75%.")
else:
    display(low_attendance.reset_index(drop=True))
```

## **Code 11: NumPy Numerical Analysis**

```python
# ================================================================
# CODE 11: NUMPY NUMERICAL ANALYSIS
# ================================================================

show_section("NUMPY NUMERICAL ANALYSIS")

marks_array = data[subjects].to_numpy(dtype=float)

print("NumPy array shape:", marks_array.shape)
print("Total subject marks:", np.sum(marks_array))
print("Mean of subject marks:", round(float(np.mean(marks_array)), 2))
print("Median of subject marks:", round(float(np.median(marks_array)), 2))
print(
    "Standard deviation:",
    round(float(np.std(marks_array)), 2)
)

numpy_subject_averages = np.mean(marks_array, axis=0)

print("\nSubject averages calculated using NumPy:")

for index in range(len(subjects)):
    print(
        f"{subjects[index]}: "
        f"{numpy_subject_averages[index]:.2f}"
    )
```

## **Code 12: Key Findings and Recommendations**

```python
# ================================================================
# CODE 12: KEY FINDINGS AND RECOMMENDATIONS
# ================================================================

show_section("KEY FINDINGS AND RECOMMENDATIONS")

print(f"1. Overall class average: {class_average:.2f}")

print(
    f"2. {total_passed} out of {total_students} students passed "
    f"all subjects."
)

print(
    f"3. Best-performing subject: {best_subject['Subject']}."
)

print(
    f"4. Lowest-average subject: {weakest_subject['Subject']}."
)

if weakest_subject["Average_Marks"] < class_average:
    print(
        f"5. Recommendation: Provide additional practice "
        f"in {weakest_subject['Subject']}."
    )
else:
    print(
        "5. Recommendation: Continue monitoring subject performance."
    )

if attendance_below_75 > 0:
    print(
        f"6. Recommendation: Follow up with "
        f"{attendance_below_75} student(s) below 75% attendance."
    )
else:
    print("6. All students have attendance of at least 75%.")
```
